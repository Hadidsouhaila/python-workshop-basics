```python
data = {
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Revenue": [100, 10, 126, 260]
}

df = pd.DataFrame(data)
df

# 2. Line plot
df.plot(x="Quarter", y="Revenue", kind="line")
plt.title("Revenue by Quarter")
plt.xlabel("Quarter")
plt.ylabel("Revenue")
plt.show()

# 3. Bar chart
df.plot(x="Quarter", y="Revenue", kind="bar")
plt.title("Revenue by Quarter")
plt.xlabel("Quarter")
plt.ylabel("Revenue")
plt.show()
# 4. Create a DataFrame for students' marks
data = {
    "Student": [
        "Amin", "Sara", "Yacine", "Layla", "Omar",
        "Nour", "Ahmad", "Meriem", "Ali", "Aya"
    ],
    "Exam Score": [54, 68, 60, 70, 77, 55, 60, 85, 90, 95]
}

df = pd.DataFrame(data)

df

# Create a histogram
df["Exam Score"].plot(kind="hist",  bins=5, figsize=(8, 5), edgecolor="black")
plt.title("Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

```python
# Load the Iris dataset from a URL
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'

iris_df = pd.read_csv(url)

# Display the first few rows
iris_df.head()

# 1. Filter virginica and display the first five rows
virginica = iris_df[iris_df['species'] == 'virginica']
virginica.head()

# 2. Find minimum and maximum petal width
iris_df['petal_width'].min()
iris_df['petal_width'].max()
# 3. Find the species with the highest average sepal length
average_sepal_length = iris_df.groupby('species')['sepal_length'].mean()
average_sepal_length
average_sepal_length.idxmax()

# 4. Scatter plot
iris_df.plot(x='petal_length', y='petal_width', kind='scatter')
plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.show()

# 5. Bar chart
average_petal_length = iris_df.groupby('species')['petal_length'].mean()
average_petal_length.plot(kind='bar')
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Length')
plt.show()
```
